# 15 Gold Chronic Disease Summary

## Purpose

This notebook creates a patient-level chronic disease analytics table.

## What We Are Doing

We will analyze patient conditions and create disease flags for major chronic disease groups.

## Why We Are Doing This

Chronic disease analytics is important for:
- population health management
- payer/provider analytics
- clinical risk stratification
- chronic care management
- ML feature engineering

## Final Output

`healthcare_catalog.gold.chronic_disease_summary`

## Final Grain

One row per patient.

## Step 1 — Import PySpark Functions

### What We Are Doing
We are importing Spark SQL functions.

### Why We Are Doing This
We need aggregation, condition matching, and feature engineering functions.

### Expected Output
Spark functions are available.

```python
from pyspark.sql.functions import *
```

## Step 2 — Read Silver Condition Table

### What We Are Doing
We are reading the clean Condition table from the Silver layer.

### Why We Are Doing This
The Condition table contains diagnosis and problem-list information.

### Expected Output
A DataFrame named `condition_df`.

```python
condition_df = spark.table(
    "healthcare_catalog.silver.condition_clean"
)

print("Condition Silver table loaded successfully.")
```

## Step 3 — Standardize Condition Text

### What We Are Doing
We are converting condition descriptions to lowercase.

### Why We Are Doing This
Lowercase text makes disease keyword matching easier and more consistent.

### Expected Output
A new column named `condition_text_lower`.

```python
condition_standardized_df = condition_df.withColumn(
    "condition_text_lower",
    lower(col("condition_description"))
)

display(condition_standardized_df)
```

## Step 4 — Create Disease Category Flags at Condition Level

### What We Are Doing
We are creating disease flags for each condition record.

### Why We Are Doing This
FHIR condition descriptions are detailed text. We need business-friendly disease categories.

### Disease Groups

- diabetes
- hypertension
- asthma
- coronary heart disease
- chronic kidney disease
- obesity
- depression/anxiety
- COPD
- stroke
- cancer

### Expected Output
Condition-level disease flags.

```python
condition_flagged_df = condition_standardized_df.withColumn(
    "diabetes_flag",
    when(col("condition_text_lower").contains("diabetes"), 1).otherwise(0)
).withColumn(
    "hypertension_flag",
    when(col("condition_text_lower").contains("hypertension"), 1).otherwise(0)
).withColumn(
    "asthma_flag",
    when(col("condition_text_lower").contains("asthma"), 1).otherwise(0)
).withColumn(
    "coronary_heart_disease_flag",
    when(
        col("condition_text_lower").contains("coronary heart disease") |
        col("condition_text_lower").contains("myocardial infarction") |
        col("condition_text_lower").contains("cardiac arrest"),
        1
    ).otherwise(0)
).withColumn(
    "chronic_kidney_disease_flag",
    when(
        col("condition_text_lower").contains("chronic kidney") |
        col("condition_text_lower").contains("renal failure"),
        1
    ).otherwise(0)
).withColumn(
    "obesity_flag",
    when(
        col("condition_text_lower").contains("obesity") |
        col("condition_text_lower").contains("body mass index 30"),
        1
    ).otherwise(0)
).withColumn(
    "mental_health_flag",
    when(
        col("condition_text_lower").contains("depression") |
        col("condition_text_lower").contains("anxiety") |
        col("condition_text_lower").contains("stress"),
        1
    ).otherwise(0)
).withColumn(
    "copd_flag",
    when(
        col("condition_text_lower").contains("copd") |
        col("condition_text_lower").contains("chronic obstructive"),
        1
    ).otherwise(0)
).withColumn(
    "stroke_flag",
    when(
        col("condition_text_lower").contains("stroke") |
        col("condition_text_lower").contains("cerebrovascular"),
        1
    ).otherwise(0)
).withColumn(
    "cancer_flag",
    when(
        col("condition_text_lower").contains("cancer") |
        col("condition_text_lower").contains("malignant") |
        col("condition_text_lower").contains("neoplasm"),
        1
    ).otherwise(0)
)

display(condition_flagged_df)
```

## Step 5 — Aggregate Disease Flags by Patient

### What We Are Doing
We are converting condition-level disease flags into patient-level disease indicators.

### Why We Are Doing This
Gold tables should be patient-level and dashboard/ML-ready.

### Expected Output
One row per patient with chronic disease flags.

```python
chronic_disease_summary_df = condition_flagged_df.groupBy(
    "patient_id"
).agg(
    max("diabetes_flag").alias("diabetes_flag"),
    max("hypertension_flag").alias("hypertension_flag"),
    max("asthma_flag").alias("asthma_flag"),
    max("coronary_heart_disease_flag").alias("coronary_heart_disease_flag"),
    max("chronic_kidney_disease_flag").alias("chronic_kidney_disease_flag"),
    max("obesity_flag").alias("obesity_flag"),
    max("mental_health_flag").alias("mental_health_flag"),
    max("copd_flag").alias("copd_flag"),
    max("stroke_flag").alias("stroke_flag"),
    max("cancer_flag").alias("cancer_flag"),
    count("*").alias("total_condition_records"),
    countDistinct("condition_description").alias("unique_condition_count")
)

display(chronic_disease_summary_df)
```

## Step 6 — Create Comorbidity Score

### What We Are Doing
We are counting how many chronic disease categories each patient has.

### Why We Are Doing This
Comorbidity burden is a common healthcare risk indicator.

### Expected Output
A new column named `chronic_disease_count`.

```python
chronic_disease_summary_df = chronic_disease_summary_df.withColumn(
    "chronic_disease_count",
    col("diabetes_flag") +
    col("hypertension_flag") +
    col("asthma_flag") +
    col("coronary_heart_disease_flag") +
    col("chronic_kidney_disease_flag") +
    col("obesity_flag") +
    col("mental_health_flag") +
    col("copd_flag") +
    col("stroke_flag") +
    col("cancer_flag")
).withColumn(
    "multi_chronic_condition_flag",
    when(col("chronic_disease_count") >= 2, 1).otherwise(0)
)

display(chronic_disease_summary_df)
```

## Step 7 — Validate Chronic Disease Prevalence

### What We Are Doing
We are calculating chronic disease prevalence across the patient population.

### Why We Are Doing This
This gives high-level population health statistics for dashboards and reporting.

### Expected Output
Population-level chronic disease prevalence metrics.

```python
display(
    chronic_disease_summary_df.select(
        sum("diabetes_flag").alias("patients_with_diabetes"),
        sum("hypertension_flag").alias("patients_with_hypertension"),
        sum("asthma_flag").alias("patients_with_asthma"),
        sum("coronary_heart_disease_flag").alias("patients_with_chd"),
        sum("chronic_kidney_disease_flag").alias("patients_with_ckd"),
        sum("obesity_flag").alias("patients_with_obesity"),
        sum("mental_health_flag").alias("patients_with_mental_health_condition"),
        sum("copd_flag").alias("patients_with_copd"),
        sum("stroke_flag").alias("patients_with_stroke"),
        sum("cancer_flag").alias("patients_with_cancer"),
        avg("chronic_disease_count").alias("avg_chronic_disease_count"),
        sum("multi_chronic_condition_flag").alias("patients_with_multiple_chronic_conditions")
    )
)
```

## Step 8 — Save Gold Chronic Disease Summary Table

### What We Are Doing
We are saving the chronic disease summary into the Gold layer.

### Why We Are Doing This
This table supports:
- chronic disease dashboards
- population health analytics
- risk stratification
- ML feature engineering

### Expected Output
A Delta table:

`healthcare_catalog.gold.chronic_disease_summary`

```python
chronic_disease_summary_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "healthcare_catalog.gold.chronic_disease_summary"
    )

print("Gold chronic_disease_summary saved successfully.")
```
