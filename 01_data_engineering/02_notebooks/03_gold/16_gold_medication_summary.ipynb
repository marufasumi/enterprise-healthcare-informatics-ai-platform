# 16 Gold Medication Summary

## Purpose

This notebook creates a patient-level medication analytics table.

## What We Are Doing

We will analyze medication usage patterns and create pharmacy-related healthcare KPIs.

## Why We Are Doing This

Medication analytics supports:
- pharmacy analytics
- chronic disease management
- polypharmacy analysis
- treatment burden analysis
- ML feature engineering
- population health management

## Final Output

`healthcare_catalog.gold.medication_summary`

## Final Grain

One row per patient.

## Step 1 — Import PySpark Functions

### What We Are Doing
We are importing Spark SQL functions.

### Why We Are Doing This
Medication analytics requires aggregation and feature engineering operations.

### Expected Output
Spark functions available.

```python
from pyspark.sql.functions import *
```

## Step 2 — Read Medication Silver Table

### What We Are Doing
We are reading the medication request table from the Silver layer.

### Why We Are Doing This
Medication requests contain prescription and treatment information.

### Expected Output
Medication DataFrame loaded successfully.

```python
medication_df = spark.table(
    "healthcare_catalog.silver.medication_request_clean"
)

print("Medication Silver table loaded successfully.")
```

## Step 3 — Standardize Medication Names

### What We Are Doing
We are converting medication names to lowercase.

### Why We Are Doing This
Lowercase text improves medication category matching consistency.

### Expected Output
A new standardized medication text column.

```python
medication_standardized_df = medication_df.withColumn(
    "medication_name_lower",
    lower(col("medication_name"))
)

display(medication_standardized_df)
```

## Step 4 — Create Medication Category Flags

### What We Are Doing
We are categorizing medications into major therapeutic groups.

### Why We Are Doing This
Medication categories support:
- pharmacy analytics
- chronic disease analysis
- treatment burden analysis
- ML feature engineering

### Medication Categories

- diabetes medications
- hypertension medications
- cholesterol medications
- pain medications
- antibiotics
- respiratory medications
- psychiatric medications
- cardiovascular medications

### Expected Output
Medication category flags.

```python
medication_flagged_df = medication_standardized_df.withColumn(
    "diabetes_medication_flag",
    when(
        col("medication_name_lower").contains("metformin") |
        col("medication_name_lower").contains("insulin"),
        1
    ).otherwise(0)
).withColumn(
    "hypertension_medication_flag",
    when(
        col("medication_name_lower").contains("lisinopril") |
        col("medication_name_lower").contains("amlodipine") |
        col("medication_name_lower").contains("hydrochlorothiazide") |
        col("medication_name_lower").contains("captopril"),
        1
    ).otherwise(0)
).withColumn(
    "cholesterol_medication_flag",
    when(
        col("medication_name_lower").contains("statin") |
        col("medication_name_lower").contains("simvastatin") |
        col("medication_name_lower").contains("atorvastatin"),
        1
    ).otherwise(0)
).withColumn(
    "pain_medication_flag",
    when(
        col("medication_name_lower").contains("ibuprofen") |
        col("medication_name_lower").contains("acetaminophen") |
        col("medication_name_lower").contains("fentanyl") |
        col("medication_name_lower").contains("oxycodone") |
        col("medication_name_lower").contains("hydrocodone"),
        1
    ).otherwise(0)
).withColumn(
    "antibiotic_flag",
    when(
        col("medication_name_lower").contains("amoxicillin") |
        col("medication_name_lower").contains("penicillin") |
        col("medication_name_lower").contains("cefuroxime") |
        col("medication_name_lower").contains("nitrofurantoin"),
        1
    ).otherwise(0)
).withColumn(
    "respiratory_medication_flag",
    when(
        col("medication_name_lower").contains("albuterol") |
        col("medication_name_lower").contains("fluticasone"),
        1
    ).otherwise(0)
).withColumn(
    "psychiatric_medication_flag",
    when(
        col("medication_name_lower").contains("methylphenidate") |
        col("medication_name_lower").contains("donepezil"),
        1
    ).otherwise(0)
).withColumn(
    "cardiovascular_medication_flag",
    when(
        col("medication_name_lower").contains("clopidogrel") |
        col("medication_name_lower").contains("nitroglycerin"),
        1
    ).otherwise(0)
)

display(medication_flagged_df)
```

## Step 5 — Aggregate Medication Metrics by Patient

### What We Are Doing
We are aggregating medication metrics at the patient level.

### Why We Are Doing This
Gold analytics tables should have one row per patient.

### Metrics Created

- total medications
- unique medications
- medication category flags
- medication burden

### Expected Output
Patient-level medication summary table.

```python
medication_summary_df = medication_flagged_df.groupBy(
    "patient_id"
).agg(

    count("*").alias("total_medication_requests"),

    countDistinct("medication_name").alias("unique_medications"),

    max("diabetes_medication_flag").alias("diabetes_medication_flag"),

    max("hypertension_medication_flag").alias("hypertension_medication_flag"),

    max("cholesterol_medication_flag").alias("cholesterol_medication_flag"),

    max("pain_medication_flag").alias("pain_medication_flag"),

    max("antibiotic_flag").alias("antibiotic_flag"),

    max("respiratory_medication_flag").alias("respiratory_medication_flag"),

    max("psychiatric_medication_flag").alias("psychiatric_medication_flag"),

    max("cardiovascular_medication_flag").alias("cardiovascular_medication_flag")
)

display(medication_summary_df)
```

## Step 6 — Create Polypharmacy Features

### What We Are Doing
We are creating medication burden and polypharmacy indicators.

### Why We Are Doing This
Polypharmacy is an important healthcare risk factor.

### Expected Output
Medication burden features.

```python
medication_summary_df = medication_summary_df.withColumn(
    "polypharmacy_flag",
    when(col("unique_medications") >= 5, 1).otherwise(0)
).withColumn(
    "high_medication_burden_flag",
    when(col("total_medication_requests") >= 10, 1).otherwise(0)
)

display(medication_summary_df)
```

## Step 7 — Validate Medication Analytics KPIs

### What We Are Doing
We are calculating pharmacy analytics KPIs.

### Why We Are Doing This
This supports:
- pharmacy analytics
- population health
- executive dashboards
- ML feature validation

### Expected Output
Population-level medication analytics summary.

```python
display(
    medication_summary_df.select(

        avg("total_medication_requests").alias("avg_total_medication_requests"),

        avg("unique_medications").alias("avg_unique_medications"),

        sum("polypharmacy_flag").alias("patients_with_polypharmacy"),

        sum("high_medication_burden_flag").alias("patients_with_high_medication_burden"),

        sum("diabetes_medication_flag").alias("patients_on_diabetes_medications"),

        sum("hypertension_medication_flag").alias("patients_on_hypertension_medications"),

        sum("cholesterol_medication_flag").alias("patients_on_cholesterol_medications"),

        sum("pain_medication_flag").alias("patients_on_pain_medications"),

        sum("antibiotic_flag").alias("patients_on_antibiotics"),

        sum("respiratory_medication_flag").alias("patients_on_respiratory_medications"),

        sum("psychiatric_medication_flag").alias("patients_on_psychiatric_medications"),

        sum("cardiovascular_medication_flag").alias("patients_on_cardiovascular_medications")
    )
)
```

## Step 8 — Save Gold Medication Summary Table

### What We Are Doing
We are saving the medication summary table into the Gold layer.

### Why We Are Doing This
This table becomes:
- Power BI source
- pharmacy analytics source
- ML feature store
- population health dataset

### Expected Output
A Delta table:

`healthcare_catalog.gold.medication_summary`

```python
medication_summary_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "healthcare_catalog.gold.medication_summary"
    )

print("Gold medication_summary saved successfully.")
```
