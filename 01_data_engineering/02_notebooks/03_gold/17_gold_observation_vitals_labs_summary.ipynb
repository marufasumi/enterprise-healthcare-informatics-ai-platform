# 17 Gold Observation Vitals Labs Summary

## Purpose

This notebook creates a patient-level vitals and labs analytics table.

## What We Are Doing

We will analyze clinical observations such as:
- BMI
- body weight
- blood pressure
- glucose
- heart rate
- respiratory rate
- oxygen saturation
- lab values

## Why We Are Doing This

Observation data is one of the most important clinical data sources for:
- population health analytics
- clinical risk stratification
- chronic disease monitoring
- predictive modeling
- Power BI dashboards

## Final Output

`healthcare_catalog.gold.observation_vitals_labs_summary`

## Final Grain

One row per patient.

## Step 1 — Import PySpark Functions

### What We Are Doing
We are importing Spark SQL functions.

### Why We Are Doing This
Observation analytics requires text matching, aggregation, and clinical feature engineering.

### Expected Output
Spark functions available.

```python
from pyspark.sql.functions import *
```

## Step 2 — Read Observation Silver Table

### What We Are Doing
We are reading the clean Observation table from the Silver layer.

### Why We Are Doing This
FHIR Observation resources contain vitals, labs, and clinical measurements.

### Expected Output
Observation DataFrame loaded successfully.

```python
observation_df = spark.table(
    "healthcare_catalog.silver.observation_clean"
)

print("Observation Silver table loaded successfully.")
```

## Step 3 — Standardize Observation Text

### What We Are Doing
We are converting observation descriptions to lowercase.

### Why We Are Doing This
Lowercase text makes clinical observation matching easier and more consistent.

### Expected Output
A new column named `observation_text_lower`.

```python
observation_standardized_df = observation_df.withColumn(
    "observation_text_lower",
    lower(col("observation_description"))
)

display(observation_standardized_df)
```

## Step 4 — Explore Common Observation Types

### What We Are Doing
We are counting the most common observation descriptions.

### Why We Are Doing This
Before feature engineering, we need to understand which vitals/labs exist in the dataset.

### Expected Output
Top observation descriptions and counts.

```python
display(
    observation_standardized_df.groupBy(
        "observation_description"
    ).count().orderBy(
        desc("count")
    ).limit(100)
)
```

## Step 5 — Create Patient-Level Vitals and Labs Summary

### What We Are Doing
We are extracting important clinical observation values using conditional aggregation.

### Why We Are Doing This
Silver observation data is long-format.  
Gold analytics needs patient-level wide-format clinical features.

### Clinical Features Created

- average BMI
- max BMI
- average systolic blood pressure
- average diastolic blood pressure
- average glucose
- average HbA1c
- average heart rate
- average respiratory rate
- average oxygen saturation
- total observation count

### Expected Output
One row per patient with clinical vitals/labs features.

```python
vitals_labs_summary_df = observation_standardized_df.groupBy(
    "patient_id"
).agg(

    count("*").alias("total_observations"),

    avg(
        when(
            col("observation_text_lower").contains("body mass index") |
            col("observation_text_lower").contains("bmi"),
            col("observation_value")
        )
    ).alias("avg_bmi"),

    max(
        when(
            col("observation_text_lower").contains("body mass index") |
            col("observation_text_lower").contains("bmi"),
            col("observation_value")
        )
    ).alias("max_bmi"),

    avg(
        when(
            col("observation_text_lower").contains("systolic blood pressure"),
            col("observation_value")
        )
    ).alias("avg_systolic_bp"),

    avg(
        when(
            col("observation_text_lower").contains("diastolic blood pressure"),
            col("observation_value")
        )
    ).alias("avg_diastolic_bp"),

    avg(
        when(
            col("observation_text_lower").contains("glucose"),
            col("observation_value")
        )
    ).alias("avg_glucose"),

    avg(
        when(
            col("observation_text_lower").contains("hemoglobin a1c") |
            col("observation_text_lower").contains("hba1c"),
            col("observation_value")
        )
    ).alias("avg_hba1c"),

    avg(
        when(
            col("observation_text_lower").contains("heart rate"),
            col("observation_value")
        )
    ).alias("avg_heart_rate"),

    avg(
        when(
            col("observation_text_lower").contains("respiratory rate"),
            col("observation_value")
        )
    ).alias("avg_respiratory_rate"),

    avg(
        when(
            col("observation_text_lower").contains("oxygen saturation"),
            col("observation_value")
        )
    ).alias("avg_oxygen_saturation")
)

display(vitals_labs_summary_df)
```

## Step 6 — Create Clinical Risk Flags

### What We Are Doing
We are creating simple clinical abnormality flags from vitals/labs.

### Why We Are Doing This
Gold tables should include business-friendly clinical indicators for dashboards and ML.

### Risk Flags Created

- obesity flag
- high blood pressure flag
- high glucose flag
- high HbA1c flag
- low oxygen flag

### Expected Output
Clinical risk flags added.

```python
vitals_labs_summary_df = vitals_labs_summary_df.withColumn(
    "obesity_from_bmi_flag",
    when(col("avg_bmi") >= 30, 1).otherwise(0)
).withColumn(
    "high_blood_pressure_flag",
    when(
        (col("avg_systolic_bp") >= 130) |
        (col("avg_diastolic_bp") >= 80),
        1
    ).otherwise(0)
).withColumn(
    "high_glucose_flag",
    when(col("avg_glucose") >= 126, 1).otherwise(0)
).withColumn(
    "high_hba1c_flag",
    when(col("avg_hba1c") >= 6.5, 1).otherwise(0)
).withColumn(
    "low_oxygen_saturation_flag",
    when(col("avg_oxygen_saturation") < 95, 1).otherwise(0)
)

display(vitals_labs_summary_df)
```

## Step 7 — Validate Vitals and Labs KPIs

### What We Are Doing
We are calculating population-level clinical metrics.

### Why We Are Doing This
This validates clinical feature engineering and supports dashboard KPIs.

### Expected Output
Population-level vitals/labs summary.

```python
display(
    vitals_labs_summary_df.select(

        avg("total_observations").alias("avg_observations_per_patient"),

        avg("avg_bmi").alias("population_avg_bmi"),

        avg("avg_systolic_bp").alias("population_avg_systolic_bp"),

        avg("avg_diastolic_bp").alias("population_avg_diastolic_bp"),

        avg("avg_glucose").alias("population_avg_glucose"),

        avg("avg_hba1c").alias("population_avg_hba1c"),

        avg("avg_heart_rate").alias("population_avg_heart_rate"),

        avg("avg_respiratory_rate").alias("population_avg_respiratory_rate"),

        avg("avg_oxygen_saturation").alias("population_avg_oxygen_saturation"),

        sum("obesity_from_bmi_flag").alias("patients_with_obesity_bmi"),

        sum("high_blood_pressure_flag").alias("patients_with_high_bp"),

        sum("high_glucose_flag").alias("patients_with_high_glucose"),

        sum("high_hba1c_flag").alias("patients_with_high_hba1c"),

        sum("low_oxygen_saturation_flag").alias("patients_with_low_oxygen")
    )
)
```

## Step 8 — Save Gold Vitals and Labs Summary Table

### What We Are Doing
We are saving the patient-level vitals/labs summary table into the Gold layer.

### Why We Are Doing This
This table becomes:
- Power BI source
- clinical analytics source
- ML feature store
- population health dataset

### Expected Output
A Delta table:

`healthcare_catalog.gold.observation_vitals_labs_summary`

```python
vitals_labs_summary_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "healthcare_catalog.gold.observation_vitals_labs_summary"
    )

print("Gold observation_vitals_labs_summary saved successfully.")
```
