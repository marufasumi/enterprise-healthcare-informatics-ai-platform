# 04 Silver Encounter Clean

## Purpose

This notebook creates the Silver Encounter table from the Bronze raw Encounter FHIR table.

## What We Are Doing

We will:
1. Read `healthcare_catalog.bronze.encounter_raw`
2. Extract useful encounter fields from nested FHIR JSON
3. Flatten encounter visit information
4. Calculate length of stay in hours
5. Save it as `healthcare_catalog.silver.encounter_clean`

## Why We Are Doing This

Encounter data represents patient visits and care events.

This table will later support:
- readmission prediction
- length-of-stay analysis
- utilization analytics
- patient risk scoring

## Expected Final Output

A clean Silver table:

`healthcare_catalog.silver.encounter_clean`

Expected columns:
- encounter_id
- patient_id
- encounter_status
- encounter_class
- encounter_start
- encounter_end
- length_of_stay_hours
- service_provider
- encounter_type

## Step 1 — Import PySpark Functions

### What We Are Doing
We are importing PySpark SQL functions.

### Why We Are Doing This
We need Spark functions to extract nested FHIR fields and calculate time differences.

### Expected Output
Spark functions will be available for this notebook.

```python
from pyspark.sql.functions import *
```

## Step 2 — Read Bronze Encounter Table

### What We Are Doing
We are reading the raw Encounter Bronze Delta table.

### Why We Are Doing This
The Bronze table contains raw nested FHIR Encounter resources.

### Expected Output
A DataFrame named `encounter_raw_df`.

```python
encounter_raw_df = spark.table("healthcare_catalog.bronze.encounter_raw")

print("Bronze encounter_raw table loaded successfully.")
```

## Step 3 — Inspect Encounter Raw Schema

### What We Are Doing
We are printing the Encounter schema.

### Why We Are Doing This
FHIR Encounter data is nested. We need to confirm where patient reference, period, status, class, and provider fields are stored.

### Expected Output
You should see nested fields such as:
- resource.id
- resource.status
- resource.class.code
- resource.subject.reference
- resource.period.start
- resource.period.end
- resource.serviceProvider.display

```python
encounter_raw_df.printSchema()
```

## Step 4 — Extract Clean Encounter Columns

### What We Are Doing

We are extracting useful encounter visit fields from nested FHIR Encounter resources.

### Why We Are Doing This

FHIR Encounter resources are deeply nested JSON structures.

Analytics and ML models require flat encounter-level tables.

This transformation creates a clean healthcare utilization dataset.

### Fields We Will Extract

- encounter_id
- patient_reference
- encounter_status
- encounter_class
- encounter_start
- encounter_end
- service_provider
- encounter_type

### Expected Output

A flattened DataFrame named:

`encounter_clean_df`

```python
from pyspark.sql.functions import col

encounter_clean_df = encounter_raw_df.select(

    col("resource.id").alias("encounter_id"),

    col("resource.subject.reference").alias("patient_reference"),

    col("resource.status").alias("encounter_status"),

    col("resource.class.code").alias("encounter_class"),

    col("resource.period.start").alias("encounter_start"),

    col("resource.period.end").alias("encounter_end"),

    col("resource.serviceProvider.display").alias("service_provider"),

    col("resource.type").alias("encounter_type")
)

display(encounter_clean_df)
```

## Step 5 — Convert Encounter Timestamps

### What We Are Doing

We are converting encounter start and end timestamps into Spark timestamp format.

### Why We Are Doing This

FHIR stores timestamps as strings.

Time-based analytics such as:
- length of stay
- visit duration
- readmission windows

require timestamp data types.

### Expected Output

Encounter timestamps converted into proper Spark timestamps.

```python
encounter_clean_df = encounter_clean_df.withColumn(
    "encounter_start",
    to_timestamp(col("encounter_start"))
)

encounter_clean_df = encounter_clean_df.withColumn(
    "encounter_end",
    to_timestamp(col("encounter_end"))
)

print("Encounter timestamps converted successfully.")
```

## Step 6 — Calculate Length of Stay

### What We Are Doing

We are calculating encounter duration in hours.

### Why We Are Doing This

Length of stay is one of the most important healthcare operational metrics.

It is commonly used for:
- hospital analytics
- readmission prediction
- ICU utilization
- cost analysis

### Expected Output

A new column:

`length_of_stay_hours`

```python
encounter_clean_df = encounter_clean_df.withColumn(
    "length_of_stay_hours",
    
    (
        unix_timestamp(col("encounter_end")) -
        unix_timestamp(col("encounter_start"))
    ) / 3600
)

print("Length of stay calculated successfully.")
```

## Step 7 — Extract Clean Patient ID

### What We Are Doing

We are extracting the actual patient ID from the FHIR patient reference field.

### Why We Are Doing This

FHIR references look like:

`urn:uuid:xxxxxxxx`

We need only the UUID portion for joins.

This patient ID will later connect:
- encounters
- observations
- conditions
- medications
- claims

### Expected Output

A new clean column:

`patient_id`

```python
encounter_clean_df = encounter_clean_df.withColumn(
    "patient_id",
    
    regexp_extract(
        col("patient_reference"),
        r'urn:uuid:(.*)',
        1
    )
)

print("Patient ID extracted successfully.")
```

## Step 8 — Inspect Clean Encounter Data

### What We Are Doing

We are displaying the flattened encounter table.

### Why We Are Doing This

We need to verify:
- nested extraction worked
- timestamps converted correctly
- LOS calculation works
- patient IDs extracted correctly

### Expected Output

A clean encounter-level healthcare table.

```python
display(encounter_clean_df)
```

```python
# debugging
encounter_raw_df.printSchema()
```

```python
encounter_clean_df = encounter_clean_df.withColumn(
    "encounter_type_text",
    regexp_extract(
        col("encounter_type"),
        '"text":"([^"]+)"',
        1
    )
)

print("encounter_type_text column created successfully.")
```

```python
display(
    encounter_clean_df.select(
        "encounter_id",
        "patient_id",
        "encounter_class",
        "encounter_type_text",
        "length_of_stay_hours"
    )
)
```

# Save Silver Encounter Table

This notebook saves the cleaned encounter-level healthcare dataset
into the Silver layer Delta table.

```python
encounter_clean_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("healthcare_catalog.silver.encounter_clean")
```
