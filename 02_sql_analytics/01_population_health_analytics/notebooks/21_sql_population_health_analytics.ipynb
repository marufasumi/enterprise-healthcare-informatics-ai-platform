# Step 13B — Learn CAST for Datatype Conversion

## What We Are Doing

In this step, we are explicitly converting the average age calculation into a decimal datatype using `CAST()`.

This helps control numeric formatting and precision.

---

## Why We Are Doing This

Different SQL engines sometimes handle numeric precision differently.

Even after using `ROUND()`, the displayed output may still show many decimal places depending on datatype behavior.

Using `CAST()` gives us more control over the output datatype.

---

## SQL Concepts Learned

This step introduces:

- CAST()
- datatype conversion
- decimal precision control

---

## What is CAST?

`CAST()` converts one datatype into another datatype.

Example:

```sql
CAST(value AS datatype)
```

---

## SQL Syntax Overview

Example:

```sql
CAST(AVG(age) AS DECIMAL(10,2))
```

Meaning:

- total digits = 10
- decimal digits = 2

---

## Expected Result

We expect the average age to display with exactly:

2 decimal places

````python
%sql

SELECT
    CAST(AVG(age) AS DECIMAL(10,2)) AS average_age

FROM healthcare_catalog.gold.population_health_dashboard;
````

# Step 46 — Learn PARTITION BY

## What We Are Doing

In this step, we are ranking healthcare patients by spending within each gender group.

---

## Why We Are Doing This

Healthcare analytics frequently compares rankings within categories.

Examples include:

- ranking patients within gender
- ranking providers within hospitals
- ranking utilization within regions

PARTITION BY creates separate ranking groups.

---

## SQL Concepts Learned

This step introduces:

- PARTITION BY
- grouped window functions
- within-group ranking

---

## Important SQL Concept

Without PARTITION BY:

Ranking occurs globally.

With PARTITION BY:

Ranking restarts within groups.

---

## SQL Structure Overview

Example:

```sql
ROW_NUMBER()

OVER(

PARTITION BY group_column

ORDER BY metric DESC

)
```

---

## Expected Result

Patients should receive rankings within their gender groups.

````python
%sql

SELECT

    patient_id,
    gender,
    age,
    total_claim_cost,

    ROW_NUMBER()

    OVER
    (
        PARTITION BY gender

        ORDER BY total_claim_cost DESC
    )

    AS gender_spending_rank

FROM healthcare_catalog.gold.patient_summary

ORDER BY gender, gender_spending_rank

````
