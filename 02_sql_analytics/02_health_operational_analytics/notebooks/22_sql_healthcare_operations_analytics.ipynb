# Query 1.1

# Population Health Risk Category vs Resource Utilization

---

## Question

Do higher-risk patients consume more healthcare encounters?

---

## What we are doing

We are grouping patients by:

population_health_risk_category

Then we calculate:

- total patients
- average encounters
- minimum encounters
- maximum encounters
- average emergency encounters
- average inpatient encounters

---

## Why we are doing it

We want to see whether higher-risk populations create more resource utilization.

In healthcare operations:

Higher Risk

↓

More Encounters

↓

More Emergency / Inpatient Use

↓

Higher Operational Burden

---

## How this works

We use:

GROUP BY

AVG()

MIN()

MAX()

COUNT()

to summarize utilization by risk category.

---

## Expected Output

Each row should show one risk category.

Example:

| population_health_risk_category | total_patients | avg_total_encounters |
|---|---:|---:|
| High Risk | xx | xx |
| Medium Risk | xx | xx |
| Low Risk | xx | xx |

---

## Interview Relevance

This is a common healthcare analytics question:

"How do you identify which patient populations consume the most resources?"

```python
%sql
SELECT

population_health_risk_category,

COUNT(*) AS total_patients,

ROUND(AVG(total_encounters), 2)
    AS avg_total_encounters,

MIN(total_encounters)
    AS min_total_encounters,

MAX(total_encounters)
    AS max_total_encounters,

ROUND(AVG(emergency_encounters), 2)
    AS avg_emergency_encounters,

ROUND(AVG(inpatient_encounters), 2)
    AS avg_inpatient_encounters

FROM healthcare_catalog.gold.population_health_dashboard

GROUP BY population_health_risk_category

ORDER BY avg_total_encounters DESC;
```

## Interview Relevance

Common healthcare analytics question:

```python
%sql
SELECT

CASE

WHEN age < 18
THEN 'Pediatric'

WHEN age BETWEEN 18 AND 64
THEN 'Adult'

ELSE 'Senior'

END AS age_group,

COUNT(*) AS total_patients,

ROUND(AVG(total_encounters),2)
AS avg_total_encounters,

ROUND(AVG(emergency_encounters),2)
AS avg_emergency_encounters,

ROUND(AVG(inpatient_encounters),2)
AS avg_inpatient_encounters

FROM healthcare_catalog.gold.population_health_dashboard

GROUP BY age_group

ORDER BY avg_total_encounters DESC;
```

# Query 3.1

# Claim Cost by Population Health Risk Category

---

## Business Question

Do high-risk populations generate greater healthcare spending?

---

## What We Are Doing

Compare:

population_health_risk_category

against:

- average claim cost
- maximum claim cost
- total claim cost

---

## Why We Are Doing It

Higher clinical risk often leads to:

More encounters

↓

More admissions

↓

More treatment

↓

Higher spending

We want to measure whether financial burden rises with risk.

---

## How This Works

We summarize:

AVG()

SUM()

MAX()

GROUP BY

---

## Expected Output

Risk groups ranked by spending.

Possible pattern:

High Risk

↓

Highest claims burden

---

## Business Value

Supports:

- cost reduction initiatives
- payer analytics
- risk stratification
- financial planning

```python
%sql
SELECT

population_health_risk_category,

COUNT(*) AS total_patients,

ROUND(
AVG(total_claim_cost),2
) AS avg_total_claim_cost,

ROUND(
SUM(total_claim_cost),2
) AS total_claim_cost_population,

ROUND(
MAX(total_claim_cost),2
) AS max_total_claim_cost

FROM healthcare_catalog.gold.population_health_dashboard

GROUP BY population_health_risk_category

ORDER BY avg_total_claim_cost DESC;
```
